```python
import os

import dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain_core.runnables import RunnableConfig
from pydantic import create_model, conlist
from langchain_tavily import TavilySearch


dotenv.load_dotenv()
```

```python
model = init_chat_model(
    model="gpt-4o-mini",
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY"),
)
```

```python
search_tool = TavilySearch(max_results=5)
```

```python
from datetime import datetime
today = datetime.now().strftime("%A, %B %d, %Y")
```

```python
agent = create_agent(
            model=model,
            tools=[search_tool],
            system_prompt=f"You are an expert content strategist focused on timely, trend-driven content ideation and planning, The current date is {today}",

        )
```

```python
user_prompt = "Bring me the events for that month."
```

```python
response = await agent.ainvoke(
            {
                "messages": [
                    {
                        "role": "user",
                        "content": user_prompt,
                    }
                ]
            },
        )
```

```python
from datetime import date, timedelta
import holidays

def upcoming_holidays(today: date, days_ahead: int = 60, country: str = "EG"):
    cutoff = today - timedelta(days=1)       # yesterday is allowed, older is not
    limit = today + timedelta(days=days_ahead)

    hol = holidays.country_holidays(
        country,
        years=[today.year, today.year + 1],  # include next year
        language="ar",
    )
    return [(d, name) for d, name in sorted(hol.items()) if cutoff <= d <= limit]

today = date.today()                          # or date(2026, 11, 16) for testing
items = upcoming_holidays(today)

text = "\n".join(f"- {d.isoformat()}: {name}" for d, name in items)
print(text or "No upcoming occasions")
```
